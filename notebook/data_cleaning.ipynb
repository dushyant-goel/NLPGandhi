```python
import re
```

```python
# Split vol 1 into pages, by character 'form-feed'
with open('../corpus/mahatma-gandhi-collected-works-volume-1.txt', 'r', encoding='utf-8') as f:
    text = f.read()

pages = text.split('\f')
```

```python
# Save individual pages.

for i, page in enumerate(pages, start=1):
    with open(f'../Vol.1: 1884 - 30 NOVEMBER, 1896/page_{i}.txt', 'w', encoding='utf-8') as out:
        out.write(page)

```

```python
# Next, remove the foot from pages,
# For odd pages
# 'VOL.1: 1884 - 30 NOVEMBER, 1896'
# <page nubmber>
# 
# For even pages
# <page number>
# THE COLLECTED WORKS OF MAHATMA GANDHI

def clean_odd_page(text):
    # Removes the 'VOL.1: ...' and then a page number at end
    text = re.sub(r'VOL\.1\: 1884 - 30 NOVEMBER, 1896\s*\n?\d+\s*$', '', text, flags=re.MULTILINE)
    return text

def clean_even_page(text):
    # Removes page number then book title at end
    text = re.sub(r'THE COLLECTED WORKS OF MAHATMA GANDHI\s*$', '', text, flags=re.MULTILINE)
    return text

```

```python
clean_pages = []
for i, page in enumerate(pages):
    if i % 2 == 0: # 0-index vs page-index
        clean_pages.append(clean_odd_page(page))
    else:
        clean_pages.append(clean_even_page(page))
```

```python
for i, clean_page in enumerate(clean_pages, start=1):
    with open(f'../Vol.1: 1884 - 30 NOVEMBER, 1896/page_{i}.txt', 'w', encoding='utf-8') as out:
        out.write(clean_page)
```
